```python
## to find sum of elements
list=[1,2,3,4,5,6,7,8]
sum=0
for i in list:
    sum=sum+i
    
print(sum)
```

```python
## to find sum of even and odd number
lst=[1,2,3,4,5,6,7]
even_sum=0
odd_sum=0

for i in lst:
    if(i%2==0):
        even_sum=even_sum+i
    else:
        odd_sum=odd_sum+i
        
print("Even sum is {}".format(even_sum))
print("odd sum is {}".format( odd_sum))
    
```
